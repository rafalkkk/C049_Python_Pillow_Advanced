In this lesson, we will learn how to load, display, and inspect images using libraries commonly used in image processing: scikit-image, imageio, and matplotlib.

#### **1. Importing images**   

There are two common functions for reading images into Python:
- `skimage.io.imread(path)` from `scikit-image`
- `imageio.imread(path)` from `imageio`

Both return the image as a NumPy array.



````python
from skimage import io
import imageio.v2 as imageio

img_skimage = io.imread('bear.jpg')
img_imageio = imageio.imread('bear.jpg')
````

````python
print(img_skimage.shape, img_imageio.shape)
````

````python
print(img_skimage[:3][:3][0])
print(img_imageio[:3][:3][0])
````

#### **2. Displaying images**

To show an image, use matplotlib’s `imshow()` function:

Let's start with something simple. We generate a matrix 2-dimentional. 
Values in the matrix can be mapped to a color using cmap

````python
import numpy as np
import matplotlib.pyplot as plt

a = np.random.random((10, 10))
print(a)
print(a.shape)
````

````python

# Display the array as an image
plt.imshow(a, cmap='viridis', interpolation='nearest')
plt.colorbar()  
plt.show()
````

The cmap can be changed to get different types of charts

````python
plt.imshow(a, cmap='gray', interpolation='nearest')
plt.colorbar()  
plt.show()
````

When working with a picture, we have already 2-dimentional map of flat space.
The values from the 3rd dimention are automatically interpreted as RGB. So, we don't need to define any cmap.

````python
import matplotlib.pyplot as plt

plt.imshow(img_skimage)
plt.title("Image loaded with skimage")
plt.axis('off')  # hide axes for clarity
plt.show()
````

#### **3. Inspecting image properties**

Each loaded image is essentially a NumPy array.
We can check its properties:
- `image.shape`: shows the dimensions (height, width, channels)
- `image.dtype`: shows the data type of pixel values

````python
print("Image shape:", img_skimage.shape)
print("Image dtype:", img_skimage.dtype)
````

If the shape is:
- **(height, width, 3)** - it means the image is RGB (3 color channels).
- **(height, width, 4)** - it includes an alpha channel (transparency).
- **(height, width)** - it is a grayscale image.
