# **Feature detection and filtering**

#### **1. Edge detection**
**Sobel Filter**
- A gradient-based edge detector.
- Computes intensity gradients in horizontal and vertical directions.
- Highlights edges by emphasizing strong transitions in intensity.

```python
from skimage import data, filters
import matplotlib.pyplot as plt

image = data.camera()
edges_sobel = filters.sobel(image)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Original image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(edges_sobel, cmap='gray')
plt.title("Edges detected with Sobel filter")
plt.axis('off')
plt.show()
```

**Canny edge detector**
- More advanced, multi-stage edge detector.
- Includes noise reduction, gradient calculation, non-maximum suppression, and hysteresis thresholding.
- Produces thin, accurate edges.

```python
from skimage.feature import canny

edges_canny = canny(image, sigma=2)

plt.imshow(edges_canny, cmap='gray')
plt.title("Edges detected with Canny")
plt.axis('off')
plt.show()
```

#### **2. Gaussian smoothing**

- Applies a Gaussian blur to reduce noise and detail.
- Controlled by standard deviation (sigma).
- Useful before edge detection to reduce false positives.

```python
from skimage.filters import gaussian

smoothed = gaussian(image, sigma=5)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Original image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(smoothed, cmap='gray')
plt.title("Gaussian smoothed image")
plt.axis('off')
plt.show()
```

#### **3. Contrast enhancement**

Histogram equalization redistributes intensity values to equalize the histogram.

```python
from skimage.exposure import equalize_hist
from skimage import io

xray_image = io.imread("xray.png")
hist_eq = equalize_hist(xray_image)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(xray_image, cmap='gray')
plt.title("Original image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(hist_eq, cmap='gray')
plt.title("Histogram equalization")
plt.axis('off')

plt.show()
```

CLAHE (Contrast Limited Adaptive Histogram Equalization) enhances contrast locally to prevent over-amplification of noise.

```python
from skimage.exposure import equalize_adapthist

clahe_img = equalize_adapthist(xray_image, clip_limit=0.03)


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(xray_image, cmap='gray')
plt.title("Original image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(clahe_img, cmap='gray')
plt.title("CLAHE")
plt.axis('off')
plt.show()
```
