# **Basic image manipulations**

#### **1. Convert to grayscale**

Use `skimage.color.rgb2gray` to convert an RGB image to grayscale (single channel).

```python
from skimage import io
from skimage.color import rgb2gray
import matplotlib.pyplot as plt
```

```python
img = io.imread('bear.jpg')

gray_image = rgb2gray(img)

plt.imshow(gray_image)
plt.axis('off')
plt.show()

print("Grayscale image shape:", gray_image.shape)  # (height, width) - single channel
```

**Why is the image colorful?**

`matplotlib’s imshow()` function uses a default colormap (often "viridis") when displaying single-channel images, mapping grayscale intensities to a spectrum of colors rather than shades of gray. To display the image in true grayscale, the `cmap='gray'` argument must be added to `imshow()`.

```python
plt.imshow(gray_image, cmap='gray')
plt.axis('off')
plt.show()
```

#### **2. Flipping images**

Flip an image horizontally or vertically using NumPy:
- `np.fliplr(image)` flips left to right (horizontal flip)
- `np.flipud(image)` flips up to down (vertical flip)



```python
import numpy as np

flipped_lr = np.fliplr(img)  # horizontal flip
flipped_ud = np.flipud(img)  # vertical flip

plt.imshow(flipped_lr)
plt.axis('off')
plt.show()

plt.imshow(flipped_ud)
plt.axis('off')
plt.show()
```

#### **3. Rotating images**

Use `skimage.transform.rotate` to rotate an image by arbitrary degrees.

```python
from skimage.transform import rotate

rotated_img = rotate(img, angle=45)

plt.imshow(rotated_img)
plt.axis('off')
plt.show()

rotated_img = rotate(img, angle=-90)

plt.imshow(rotated_img)
plt.axis('off')
plt.show()
```

Rotation may change the image size or introduce empty corners (black regions) if not cropped.

#### **4. Resizing and rescaling**

Two common methods:
- `resize(image, output_shape)` resizes to exact dimensions.
- `rescale(image, scale)` rescales by a factor (e.g., 0.5 halves the size).

```python
from skimage.transform import resize, rescale

resized_img = resize(img, (300, 300), anti_aliasing=True)

plt.imshow(resized_img)
plt.axis('off')
plt.show()
```

**Aliasing and anti-aliasing**

Downscaling images may cause aliasing artifacts.
Using `anti_aliasing=True` helps reduce these artifacts.

Let's look at the upper example but this time using `anti_aliasing=False`.

```python
rescaled_img = rescale(img, 0.1, anti_aliasing=False, channel_axis=-1)

plt.imshow(rescaled_img)
plt.axis('off')
plt.show()
```

Rescale does a simmilar thing, but the new size is given by scale (not by exact size)

```python
rescaled_img = rescale(img, 0.1, anti_aliasing=True, channel_axis=-1)

plt.imshow(rescaled_img)
plt.axis('off')
plt.show()
```

#### **5. Region of interest (ROI)**

Crop a rectangular region from an image.

```python
roi = img[60:450, 410:850]

plt.imshow(roi) 
plt.axis('off')
plt.show()
```

#### **6. Saving a modified image**

```python
io.imsave('bear_head.png', roi)
```

```python
img2 = io.imread('bear_head.png')

plt.imshow(img2)
plt.axis('off')
plt.show()
```
